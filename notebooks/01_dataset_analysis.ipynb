```python
import os
import shutil
from sklearn.model_selection import train_test_split
```

```python
raw_train_dir = "../data/raw/archive/Training"
raw_test_dir = "../data/raw/archive/Testing"

processed_dir = "../data/processed"

train_dir = os.path.join(processed_dir, "train")
val_dir = os.path.join(processed_dir, "val")
test_dir = os.path.join(processed_dir, "test")
```

```python
classes = ["glioma", "meningioma", "pituitary", "notumor"]

for split_dir in [train_dir, val_dir, test_dir]:
    for cls in classes:
        os.makedirs(os.path.join(split_dir, cls), exist_ok=True)

print("Folders created successfully.")
```

```python
for cls in classes:
    
    class_path = os.path.join(raw_train_dir, cls)
    
    images = os.listdir(class_path)
    
    train_images, val_images = train_test_split(
        images,
        test_size=0.2,
        random_state=42
    )
    
    # Copy training images
    for img in train_images:
        src = os.path.join(class_path, img)
        dst = os.path.join(train_dir, cls, img)
        shutil.copy(src, dst)
    
    # Copy validation images
    for img in val_images:
        src = os.path.join(class_path, img)
        dst = os.path.join(val_dir, cls, img)
        shutil.copy(src, dst)

print("Training and validation split completed.")
```

```python
for cls in classes:
    
    class_path = os.path.join(raw_test_dir, cls)
    
    images = os.listdir(class_path)
    
    for img in images:
        src = os.path.join(class_path, img)
        dst = os.path.join(test_dir, cls, img)
        shutil.copy(src, dst)

print("Test data copied successfully.")
```

```python
for split in ["train", "val", "test"]:
    
    print(f"\n{split.upper()} SET")
    
    split_path = os.path.join(processed_dir, split)
    
    for cls in classes:
        
        count = len(os.listdir(os.path.join(split_path, cls)))
        
        print(f"{cls}: {count}")
```
